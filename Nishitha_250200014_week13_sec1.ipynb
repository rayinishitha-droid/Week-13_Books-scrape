```python
book=soup
```

```python
title = book.h3.a["title"]
price = book.find("p", class_="price_color").text

# Extract rating
rating_class = book.p["class"][1]

print(f"Title: {title}")
print(f"Price: {price}")
print(f"Rating: {rating_class} stars")
```

```python
all_books=soup.find_all('li',class_='col-xs-6 col-sm-4 col-md-3 col-lg-3')
```

```python
books_data = []
# Loop through each book and extract details
for book in all_books:
    title = book.h3.a["title"]
    price = book.find("p", class_="price_color").text

    # Extract rating class name (e.g., ['star-rating', 'Three'])
    rating_class = book.p["class"][1]

    book_info = {
        "title": title,
        "price": price,
        "rating": f"{rating_class} stars"
    }

    books_data.append(book_info)
```

```python
books_data
```

```python
len(books_data)
```

```python
import pandas as pd
df=pd.DataFrame(books_data)
df.head()
```

```python
df.info()
```

```python
df.to_csv("books_1stPage.csv",index=False)
```
